```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

pdf_path = "2410.10934v2_7033 1.pdf"

loader = PyPDFLoader(pdf_path)
documents = loader.load()

print(f"Number of pages loaded: {len(documents)}")
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=700,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documents)

print(f"Number of chunks created: {len(chunks)}")
```

```python
print(chunks[0].page_content)
```

```python
print("\nMetadata:")
print(chunks[0].metadata)
```

```python
from collections import Counter

page_counts = Counter(
    chunk.metadata["page"]
    for chunk in chunks
)

print(page_counts)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="BAAI/bge-base-en-v1.5"
)

print("Embedding model loaded")
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)
vectorstore.save_local("faiss_index")

print("FAISS index saved successfully!")
print("FAISS vector store created successfully!")
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.4
)
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 4}
)

print("Retriever created successfully!")
```

```python
query = "What is Agent-as-a-Judge?"

retrieved_docs = retriever.invoke(query)

print(f"Retrieved {len(retrieved_docs)} chunks\n")

for i, doc in enumerate(retrieved_docs, 1):
    print(f"--- Chunk {i} ---")
    print(doc.page_content[:1000])
    print(f"Page: {doc.metadata.get('page')}")
    print()
```

```python

retriever = vectorstore.as_retriever(
    search_kwargs={"k": 4}
)

print("Retriever created successfully!")
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are answering questions about the provided document.

Use ONLY the context provided below to answer the question.
If the answer cannot be found in the context, say:
"I could not find the answer in the provided document."

Context:
{context}

Question:
{question}

Answer:
""")

print("RAG prompt created successfully!")
```

```python
question = "What is the DevAI dataset, and how many tasks, requirements, and preferences does it contain?"

retrieved_docs = retriever.invoke(question)

context = "\n\n".join(
    doc.page_content for doc in retrieved_docs
)

messages = prompt.invoke({
    "context": context,
    "question": question
})

response = llm.invoke(messages)

print("ANSWER")
print("=" * 80)
print(response.content)

print("\nSOURCES")
print("=" * 80)

for i, doc in enumerate(retrieved_docs, 1):
    page = doc.metadata.get("page", None)

    if page is not None:
        page = page + 1

    print(f"\nSource {i}")
    print(f"Page: {page}")
    print("-" * 80)
    print(doc.page_content)
```

```python
import ragas

print("RAGAS version:", ragas.__version__)

from ragas.metrics import (
    Faithfulness,
    ResponseRelevancy,
    ContextPrecision,
    ContextRecall
)

print("RAGAS metrics imported successfully!")
```

```python
question = "What is the DevAI dataset, and how many tasks, requirements, and preferences does it contain?"

# Run your existing RAG pipeline
retrieved_docs = retriever.invoke(question)

context = [
    doc.page_content
    for doc in retrieved_docs
]

messages = prompt.invoke({
    "context": "\n\n".join(context),
    "question": question
})

response = llm.invoke(messages)

answer = response.content

print("Question:")
print(question)

print("\nGenerated Answer:")
print(answer)

print("\nNumber of Retrieved Contexts:")
print(len(context))
```

```python
from ragas import EvaluationDataset, SingleTurnSample

reference_answer = """
Agent-as-a-Judge is a framework in which agentic systems are used
to evaluate other agentic systems. It extends the LLM-as-a-Judge
approach and evaluates agent performance while considering the
step-by-step nature of agentic systems.
"""

sample = SingleTurnSample(
    user_input=question,
    response=answer,
    retrieved_contexts=context,
    reference=reference_answer
)

dataset = EvaluationDataset(
    samples=[sample]
)

print("RAGAS evaluation dataset created!")
```

```python
from ragas.llms import LangchainLLMWrapper

ragas_llm = LangchainLLMWrapper(llm)

print("RAGAS evaluator LLM configured!")
```

```python
faithfulness = Faithfulness(llm=ragas_llm)

response_relevancy = ResponseRelevancy(
    llm=ragas_llm,
    embeddings=embeddings
)

context_precision = ContextPrecision(llm=ragas_llm)

context_recall = ContextRecall(llm=ragas_llm)

print("RAGAS metrics configured successfully!")
```

```python
from ragas import evaluate

result = evaluate(
    dataset=dataset,
    metrics=[
        faithfulness,
        context_precision,
    ]
)

print(result)
```
